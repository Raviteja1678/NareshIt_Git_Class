## Numpy

- Numpy means : **`Numerical Python`** or **Number Python**

- **All the `math Operations` ==== done by **Numpy Package**

- A student should understand the number representations

    - **Basic Level** : **`List and tuple`**
 
    - list and tuple are not **vector representation**
 
    - so thats why === `NumPy`


- In Numpy we will **represent the numbers** === called as **`Array`**
 
- `Array` means == a List of numbers, But `not vectors`

- Same like Numpy === there is one more `Math package` : **`tensorflow`**


#### for CPU ==== NumPy arrays are Best

#### for TPU ==== tensorflow tensors are Best

```python
[10,20,30,45]       # scalar data

array([10,20,30,45])   # vector used in  Machine Learning

tensor([10,20,30,45])     # tensor (Google )  used in DeepLearning
```

```python
Examples
--------
>>> import numpy as np
>>> np.array([1, 2, 3])
array([1, 2, 3])

Upcasting:

>>> np.array([1, 2, 3.0])
array([ 1.,  2.,  3.])

More than one dimension:

>>> np.array([[1, 2], [3, 4]])
array([[1, 2],
       [3, 4]])

Minimum dimensions 2:

>>> np.array([1, 2, 3], ndmin=2)
array([[1, 2, 3]])

Type provided:

>>> np.array([1, 2, 3], dtype=complex)
array([ 1.+0.j,  2.+0.j,  3.+0.j])

Data-type consisting of more than one element:

>>> x = np.array([(1,2),(3,4)],dtype=[('a','<i4'),('b','<i4')])
>>> x['a']
array([1, 3], dtype=int32)

Creating an array from sub-classes:

>>> np.array(np.asmatrix('1 2; 3 4'))
array([[1, 2],
       [3, 4]])

>>> np.array(np.asmatrix('1 2; 3 4'), subok=True)
matrix([[1, 2],
        [3, 4]])
```

### can't multiply lists

```python
l1*l2
```

### with array ==== all possible 

- but it should have same no.of elements

```python
arr1*arr2
```

```python
ans = []
for i,j in zip(l1,l2):
    ans.append([i,j])
```
