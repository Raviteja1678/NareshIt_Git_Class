## upcasting

- if we element as other datatype ===
    - all others also changed to that type

```python
l1 = [1,2,3,4]
l2 = [10,20,30,40.0]    # 40.0 float
data = [l1,l2]
arr=np.array(data)
arr                       # all converted to float
```

```python
arr.dtype      #float  64 == o.s
```

```python
l1 = [1,2,3,4]
l2 = [10,20,30,40]
l3 = ['A','B','C','D']
data = [l1,l2,l3]
arr=np.array(data)
print(data)
print(arr)
```

```python
arr.size, arr.shape, arr.ndim
```

## dimensions

```python
l1 = [1,2,3,4]
print(np.array(l1))   # [1]
print(np.array([l1]).ndim)    # [[1]]
print(np.array([[l1]]).ndim)   # [[[1]]]
```
